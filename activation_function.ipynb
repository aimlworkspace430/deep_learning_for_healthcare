```python
import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-10, 10, 500)

# Sigmoid function
sigma = 1 / (1 + np.exp(-x))

# Derivative of sigmoid
sigma_derivative = sigma * (1 - sigma)

# Plot sigmoid
plt.figure(figsize=(8, 5))
plt.plot(x, sigma)
plt.xlabel("x")
plt.ylabel("σ(x)")
plt.title("Sigmoid Function")
plt.grid(True)
plt.show()

# Plot derivative
plt.figure(figsize=(8, 5))
plt.plot(x, sigma_derivative)
plt.xlabel("x")
plt.ylabel("σ'(x)")
plt.title("Derivative of Sigmoid Function")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Original input
x = np.linspace(-10, 10, 500)

# Sigmoid output: y = sigma(x)
y = 1 / (1 + np.exp(-x))

# Log terms
log_sigma = np.log(y)
log_one_minus_sigma = np.log(1 - y)

# Plot with sigmoid output y on x-axis
plt.figure(figsize=(8, 5))

plt.plot(y, log_sigma, label="log(σ)")
plt.plot(y, log_one_minus_sigma, label="log(1 - σ)")

plt.xlabel("y = σ(x)")
plt.ylabel("Log value")
plt.title("log(σ) and log(1 - σ)")
plt.grid(True)
plt.legend()

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-10, 10, 500)

# ReLU
relu = np.maximum(0, x)

# Derivative of ReLU
relu_derivative = np.where(x < 0, 0, 1)

# Plot both
plt.figure(figsize=(8, 5))

plt.plot(x, relu, label="ReLU(x)")
plt.plot(x, relu_derivative, label="ReLU'(x)")

plt.xlabel("x")
plt.ylabel("Value")
plt.title("ReLU Function and Its Derivative")
plt.grid(True)
plt.legend()

plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# x values
x = np.linspace(-10, 10, 500)

# Leaky ReLU parameter
alpha = 0.2

# Leaky ReLU
leaky_relu = np.where(x < 0, alpha * x, x)

# Derivative of Leaky ReLU
leaky_relu_derivative = np.where(x < 0, alpha, 1)

# Plot both
plt.figure(figsize=(8, 5))

plt.plot(x, leaky_relu, label="Leaky ReLU(x)")
plt.plot(x, leaky_relu_derivative, label="Leaky ReLU'(x)")

plt.xlabel("x")
plt.ylabel("Value")
plt.title("Leaky ReLU and Its Derivative")
plt.grid(True)
plt.legend()

plt.show()
```
